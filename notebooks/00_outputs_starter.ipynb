# Starter: read runs from `outputs/`

Every run of `main.py` is written by Hydra to
`outputs/<experiment_name>/<YYYY-MM-DD_HH-MM-SS>/` with:

- `exp/`: raw data (`checkpoint.npz`, `nqs_variables.msgpack`),
- `logs/`: execution log (`main.log`),
- `plots/`: figures written by `QRC.NQS.utils.makeplots`,
- `.hydra/`: the resolved config of the run.

Create a run first, from the repository root:

```bash
uv run python main.py experiment_name=default
```

This notebook finds the repository root on its own, so it works from `notebooks/`.

````python
from QRC.NQS.utils.outputs import (
    find_repo_root,
    latest_run,
    list_experiments,
    list_runs,
    load_checkpoint,
)

ROOT = find_repo_root()
OUTPUTS = ROOT / "outputs"
EXPERIMENT = "default"

print("outputs:", OUTPUTS)
print("experiments:", list_experiments(OUTPUTS))
print("runs of", EXPERIMENT, ":", [r.root.name for r in list_runs(EXPERIMENT, OUTPUTS)])
````

````python
run = latest_run(EXPERIMENT, OUTPUTS)
weights, metadata = load_checkpoint(run.exp / "checkpoint.npz")
print("run:", run.root)
metadata
````

````python
from IPython.display import Image, display

for figure in sorted(run.plots.glob("*.png")):
    print(figure.name)
    display(Image(filename=str(figure)))
````
